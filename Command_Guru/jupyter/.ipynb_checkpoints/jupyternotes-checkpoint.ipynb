# **Setup**

## *Install*

jupyter lab --version
--If you get a version number, you're ready

pip install jupyterlab --if not installed

ls ~/.venvs/jupyter/bin/ | grep jupyter
--If you see things like:
    -jupyter
    -jupyter-lab
    -jupyter-notebook
  --activate that environment in fish:
    source ~/.venvs/jupyter/bin/activate.fish
    jupyter lab
    http://localhost:8888/lab
