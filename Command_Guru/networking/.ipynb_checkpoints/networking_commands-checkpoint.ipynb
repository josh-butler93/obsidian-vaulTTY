### *Networking Scripts*

```python
import ipaddress

net = ipaddress.ip_network("192.168.10.0/27")

print("Network:", net.network_address)
print("Broadcast:", net.broadcast_address)
print("Usable hosts:", net.num_addresses - 2)
```

```python
def wildcard(mask):
    parts = [255 - int(x) for x in mask.split(".")]
    return ".".join(map(str, parts))

wildcard("255.255.255.0")
```
