### *Git Config Management*

mkdir git-config-lab
cd git-config-lab
git init
git config --list
git config user.name
git config --global user.name "Your Name"
git config --global user.email "jane.doe@example.com"The
    --global flag: tells your time machine to apply these settings to all your time travel experiments on this systemgit config
    --global user.name
git config --global user.email
git config --global color.ui auto
    --git config --global color.ui
        -output should be auto
git config --global core.editor nano
    --git config --global core.editor
        -You should see nano as the output
git config --global core.autocrlf input 
    --This tells your time machine to convert CRLF to LF when you commit (add events to your timeline), but not the other way around when you checkout (view events from the timeline)
git config --global alias.st status
    --This creates an alias st for the status command
    --Now, instead of typing git status, you can simply type git st
            ** git st
 git config --global alias.lg "log --color --graph --pretty=format:'%Cred%h%Creset -%C(yellow)%d%Creset %s %Cgreen(%cr) %C(bold blue)<%an>%Creset' --abbrev-commit"    
    --What it does is create an alias lg that shows your time travel history in a colorful and informative way
git config --global alias.st
git config --global alias.lg
git config user.name "Lab User"
    --Notice that we didn't use the --global flag this time
