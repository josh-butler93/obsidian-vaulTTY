## **Git**

### *Cloning a Repository*

git clone http://gitea:3000/sarah/story-blog.git
git log --may allow you to see the logs of the repo
git remote -v --run this command to see if you can push to the repo

### *Creating a New Branch*

*mkdir git-branch-lab*
*cd git-branch-lab*
*git init*
*git status*
**git branch <branch_name>** #this just creates the branch
    --run git switch <branch_name> to switch to the branch
**git checkout -b <new_branch_name>**
**git check**

### *Amending a Commit*

*git add hello.txt*
**git commit --amend -m "Initial commit with important note"** ==Instead of creating a new commit in your Git history, Git has effectively replaced the previous "Initial commit" with a new, improved version
**git log --oneline**
Only Amend Unpushed Commits

### *Reverting a Commit*

**git reset**
  this can be destructive, especially if you've already pushed your changes or if others are working on the same branch
**git revert**
