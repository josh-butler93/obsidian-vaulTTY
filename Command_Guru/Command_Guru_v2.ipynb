### **Cloning a Repository**

### **Creating a New Branch**

--mkdir git-branch-lab
--cd git-branch-lab
--git init
--git status
--git branch <branch_name> #this just creates the branch
    --run git switch <branch_name> to switch to the branch
--git checkout -b <new_branch_name>
--git check
