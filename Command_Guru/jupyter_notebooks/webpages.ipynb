#### Networking calculator

````python
import ipaddress

network_input = "192.168.10.34/27"

interface = ipaddress.ip_interface(network_input)
network = interface.network

hosts = list(network.hosts())

wildcard = ".".join(
    str(255 - int(octet))
    for octet in str(network.netmask).split(".")
)

print("IP Address:     ", interface.ip)
print("CIDR:           ", network.prefixlen)
print("Network:        ", network.network_address)
print("Subnet Mask:    ", network.netmask)
print("Wildcard Mask:  ", wildcard)
print("Broadcast:      ", network.broadcast_address)
print("First Host:     ", hosts[0])
print("Last Host:      ", hosts[-1])
print("Usable Hosts:   ", len(hosts))
````

#### Homelab Memory Chart

````python
import json
import subprocess
from pathlib import Path
import matplotlib.pyplot as plt

cmd = [
    "ansible",
    "all",
    "-i",
    "/home/sysadmin/Documents/ansible/inventory.ini",
    "-m",
    "setup",
    "-a",
    "filter=ansible_hostname,ansible_memtotal_mb",
    "--tree",
    "/tmp/ansible-facts"
]

subprocess.run(cmd, capture_output=True, text=True)

hosts = []
memory_gb = []

for file in Path("/tmp/ansible-facts").iterdir():
    data = json.loads(file.read_text())
    facts = data["ansible_facts"]

    hosts.append(
        facts.get("ansible_hostname", file.name)
    )

    memory_gb.append(
        round(facts.get("ansible_memtotal_mb", 0) / 1024, 1)
    )

plt.figure(figsize=(8, 4))
plt.bar(hosts, memory_gb)

plt.title("Homelab Memory by Host")
plt.xlabel("Host")
plt.ylabel("Memory (GB)")
plt.xticks(rotation=30)
plt.tight_layout()

plt.show()
````

#### Randonmized Quiz

````python
import random

questions = [
    {
        "question": "What port does SSH use?",
        "answer": "22"
    },
    {
        "question": "What port does HTTPS use?",
        "answer": "443"
    },
    {
        "question": "What command displays the Linux routing table?",
        "answer": "ip route"
    },
    {
        "question": "What Git command shows repository status?",
        "answer": "git status"
    },
    {
        "question": "What kubectl command lists pods?",
        "answer": "kubectl get pods"
    },
    {
        "question": "What Ansible module tests connectivity?",
        "answer": "ping"
    }
]

random.shuffle(questions)

score = 0

for number, item in enumerate(questions, start=1):

    print(f"\nQuestion {number}/{len(questions)}")
    print(item["question"])

    response = input("Answer: ").strip().lower()

    if response == item["answer"].lower():
        print("✓ Correct!")
        score += 1
    else:
        print(f"✗ Incorrect — Answer: {item['answer']}")

print("\n--------------------")
print(f"Final Score: {score}/{len(questions)}")
print(f"Percentage: {(score / len(questions)) * 100:.0f}%")
````

````python
import random
import ipywidgets as widgets
from IPython.display import display, clear_output

questions = [
    ("What port does SSH use?", "22"),
    ("What port does HTTPS use?", "443"),
    ("What command displays the Linux routing table?", "ip route"),
    ("What Git command shows repository status?", "git status"),
    ("What kubectl command lists pods?", "kubectl get pods"),
    ("What Ansible module tests connectivity?", "ping"),
]

random.shuffle(questions)

current = 0
score = 0

question_text = widgets.HTML()
answer_box = widgets.Text(
    placeholder="Type your answer here..."
)
submit_button = widgets.Button(
    description="Submit Answer",
    button_style="primary"
)
feedback = widgets.HTML()
score_display = widgets.HTML()

def show_question():
    if current < len(questions):
        question_text.value = (
            f"<h3>Question {current + 1} of {len(questions)}</h3>"
            f"<p><b>{questions[current][0]}</b></p>"
        )
        answer_box.value = ""
        feedback.value = ""
        score_display.value = f"<b>Score:</b> {score}/{current}"
    else:
        percentage = (score / len(questions)) * 100

        question_text.value = "<h2>Quiz Complete 🎉</h2>"
        feedback.value = (
            f"<h3>Final Score: {score}/{len(questions)}</h3>"
            f"<p>{percentage:.0f}%</p>"
        )

        answer_box.layout.display = "none"
        submit_button.layout.display = "none"
        score_display.value = ""

def check_answer(_):
    global current, score

    expected = questions[current][1]
    response = answer_box.value.strip()

    if response.lower() == expected.lower():
        score += 1
        feedback.value = "<p>✅ <b>Correct!</b></p>"
    else:
        feedback.value = (
            f"<p>❌ <b>Incorrect.</b> "
            f"Correct answer: <code>{expected}</code></p>"
        )

    current += 1

    # Briefly show feedback, then move when button is clicked again
    submit_button.description = "Next Question"
    submit_button.on_click(next_question, remove=True)

def next_question(_):
    submit_button.description = "Submit Answer"
    submit_button.on_click(check_answer, remove=True)
    show_question()

submit_button.on_click(check_answer)

display(
    widgets.VBox([
        question_text,
        answer_box,
        submit_button,
        feedback,
        score_display
    ])
)

show_question()
````

#### Ansible Dashboard

````python
import json
import subprocess
from IPython.display import HTML, display

cmd = [
    "ansible",
    "all",
    "-i",
    "/home/sysadmin/Documents/ansible/inventory.ini",
    "-m",
    "setup",
    "-a",
    "filter=ansible_hostname,ansible_distribution,ansible_distribution_version,ansible_default_ipv4,ansible_memtotal_mb,ansible_processor_vcpus",
    "--tree",
    "/tmp/ansible-facts"
]

subprocess.run(cmd, capture_output=True, text=True)

hosts = []

from pathlib import Path

for file in Path("/tmp/ansible-facts").iterdir():
    data = json.loads(file.read_text())
    facts = data["ansible_facts"]

    hosts.append({
        "host": facts.get("ansible_hostname", file.name),
        "os": f"{facts.get('ansible_distribution', 'Unknown')} {facts.get('ansible_distribution_version', '')}",
        "ip": facts.get("ansible_default_ipv4", {}).get("address", "N/A"),
        "cpu": facts.get("ansible_processor_vcpus", "N/A"),
        "memory": round(facts.get("ansible_memtotal_mb", 0) / 1024, 1),
    })

rows = ""

for host in hosts:
    rows += f"""
    <tr>
        <td>{host['host']}</td>
        <td>{host['os']}</td>
        <td>{host['ip']}</td>
        <td>{host['cpu']}</td>
        <td>{host['memory']} GB</td>
    </tr>
    """

html = f"""
<div style="
    border: 1px solid #555;
    border-radius: 12px;
    padding: 18px;
    font-family: sans-serif;
">

<h2>🖥️ Homelab Systems</h2>

<table style="
    width: 100%;
    border-collapse: collapse;
">

<tr>
    <th style="text-align:left; padding:8px; border-bottom:1px solid #666;">Host</th>
    <th style="text-align:left; padding:8px; border-bottom:1px solid #666;">OS</th>
    <th style="text-align:left; padding:8px; border-bottom:1px solid #666;">IP Address</th>
    <th style="text-align:left; padding:8px; border-bottom:1px solid #666;">vCPU</th>
    <th style="text-align:left; padding:8px; border-bottom:1px solid #666;">Memory</th>
</tr>

{rows}

</table>

</div>
"""

display(HTML(html))
````
