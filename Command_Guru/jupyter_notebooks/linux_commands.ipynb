# **Definiitions**

#### **sed**
A command-line tool for editing text

#### **sort**
Places identical hostnames next to each other


# **Commands**

#### **grep**

***cat /etc/os-release | grep -i os***
**grep -n 'replicas:' /home/labex/project/scale-lab/hostname-web.yaml**
    - -i means case-insensitive
    - -E enables extended regular expressions, which is what lets you easily search for multiple alternatives using |


#### **sed**

***sed -i 's/nginx:1.27-alpine-missing/nginx:1.27-alpine/' broken-web.yaml***
    - sed: A command-line tool for editing text.
    - -i: Edits the file directly, rather than only displaying the modified content
    - 's/old/new/': The substitution pattern:
        - s means “substitute.”
        - nginx:1.27-alpine-missing is the text to find
        - nginx:1.27-alpine is the replacement text
        - broken-web.yaml: The file being edited

#### **sort / uniq -c**

***sort /tmp/hostname-responses.txt | uniq -c***
**sort -u /tmp/hostname-responses.txt | wc -l**
    - sort places identical hostnames next to each other
    - | sends the sorted output to the next command
    - uniq -c combines adjacent duplicate lines and prefixes each hostname with its count  
    - -u means “unique,” so duplicate hostnames are removed

#### **wc -l**

**sort -u /tmp/hostname-responses.txt | wc -l**
    - wc counts text
    - -l tells it to count lines
