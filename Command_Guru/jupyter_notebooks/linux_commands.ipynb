# **Definiitions**

#### **sed**
A command-line tool for editing text

#### **sort**
Places identical hostnames next to each other

#### **setgid bit**
When this is enabled, all new files and subdirectories inherit the directory's group ownership instead of the creator's primary group

#### **sticky bit**
With the sticky bit enabled, users can delete only files they own, even if they have write permission to the directory
*Normally, users with write permission to a directory can delete any file inside it*

#### **wc**
command will print just the number of lines, words, and characters in the output

**ps aux | grep ^root** only return processes that start with root
**ps aux | grep d$** --only return characters that end with the letter d
    - -i means case-insensitive
    - -E enables extended regular expressions, which is what lets you easily search for multiple alternatives using |
    - -n includes the matching line number
    - 'replicas:' is the text being searched for
**grep copyright** --cd /usr/share/doc
*Running grep copyright to search for all files containing the word copyright will fail*
*This happens because grep, when it's run like this, expects searchable text via standard input (stdin) - but this command doesn't provide any. You can escape the failed command by typing CTRL+c*
**grep -r copyright**
*the -r flag to the grep command so it will search recursively for all the files in the current directory tree that contain the text copyright*
***grep -rl copyright***
*Add the l flag to that command if you want grep to print nothing more than the directory and file names*
**grep -r copyright | wc**
**grep -R copyright | wc**
    - the uppercase -R flag instead of the lowercase -r will recursively return even objects that are symbolic links
**cat git/README.source | grep -A 2 extracted**
*Add the -A flag to grep with the value of 2 to print the two lines that come after the matching line*
**cat git/README.source | grep -B 5 extracted**
*If you need to see the lines that come before the matching line, use the -B flag along with some value instead of -A*
**cat git/README.source | grep -C 4 extracted**
*If you want to see both lines before and after the matching line, use the -C (context) flag*

#### ps aux

#### sed

***sed -i 's/nginx:1.27-alpine-missing/nginx:1.27-alpine/' broken-web.yaml***
    - sed: A command-line tool for editing text.
    - -i: Edits the file directly, rather than only displaying the modified content

        - s means “substitute.”
        - nginx:1.27-alpine-missing is the text to find
        - nginx:1.27-alpine is the replacement text
        - broken-web.yaml: The file being edited

#### sort / uniq -c

***sort /tmp/hostname-responses.txt | uniq -c***
**sort -u /tmp/hostname-responses.txt | wc -l**
    - sort places identical hostnames next to each other
    - | sends the sorted output to the next command
    - uniq -c combines adjacent duplicate lines and prefixes each hostname with its count  
    - -u means “unique,” so duplicate hostnames are removed

#### wc -l

**sort -u /tmp/hostname-responses.txt | wc -l**
    - wc counts text
    - -l tells it to count lines
