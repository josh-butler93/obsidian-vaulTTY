Sed
-----

* **sed -i 's/nginx:1.27-alpine-missing/nginx:1.27-alpine/' broken-web.yaml**
    - sed: A command-line tool for editing text.
    - -i: Edits the file directly, rather than only displaying the modified content
    - 's/old/new/': The substitution pattern:
        - s means “substitute.”
        - nginx:1.27-alpine-missing is the text to find
        - nginx:1.27-alpine is the replacement text
        - broken-web.yaml: The file being edited
