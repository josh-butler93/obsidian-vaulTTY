#### **Keyboard Shortcuts VSCode Edition**
* The default keyboard shortcut to toggle the visibility of the secondary sidebar (also known as the side panel) in Visual Studio Code depends on your operating system:Windows / Linux: Ctrl + Alt + B
* The default keyboard shortcut to toggle the visibility of the primary sidebar (the main panel on the left side containing File Explorer, Search, Source Control, etc.) in Visual Studio Code is:Windows / Linux: Ctrl + B

#### ** Collapsing Markdown Sections (Headings)**
* If you organize your notebook with markdown headers (e.g., # Heading 1), you can collapse everything nested under that heading:
* Toggle Current Heading: Press Left Arrow (to collapse)
* Right Arrow (to expand) while the heading cell is selected in Command Mode
* Collapse All Headings: Use Ctrl + Shift + Left Arrow.Expand All Headings: Use Ctrl + Shift + Right Arrow
