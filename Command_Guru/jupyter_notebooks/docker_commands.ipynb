# **Definitions**

**docker command**
is the client

**Docker Engine**
service is the server
*Together these form a client-server architecture*

**-p HOST_PORT:CONTAINER_PORT**
The left side (before the colon) specifies the host port, the right side (after the colon) is the container port

**Dockerfile**
is comprised of instructions
The first instruction is usually FROM which specifies the base of a new image
You will be modifying the nginx image, so this is the base image
If you were modifying the Jenkins image you would start with FROM jenkins/jenkins instead

**Dockerfile**
is a text document that contains all the commands you could call on the command line to assemble an image

# **Commands**

#### Pushing images to Docker Hub
*Eventually you will want to use images to deploy an application. This is where pushing images comes into play. - First, run docker login to provide credentials to Docker Hub, or a private registry. - Then, docker image push api pushes your image to the registry. - Finally, docker image pull api on another machine and that's it*
**docker login**
**docker image push api** --pushes image created to my image repo
**docker image pull api** --pull your new image
