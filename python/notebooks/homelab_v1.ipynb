## VSCode Code Space setup

### Setup

Similar to google colab vscode has a codespace that you can use in the browser to spin up a dev environment

Video Link: https://www.youtube.com/watch?v=fnuVw9TeYmA

1. Go to the dashboard
2. Click Codespace
3. Open the repo you want to do coding in and select how many cores you want for the machine
4. Once in there install the jupyter extension as well as the google colab extension as well --ensure it is the extension from google

Codespaces can use a config file called:
                                                                                                  
      ─ text           
      .devcontainer/devcontainer.json                                                             
                                                                                                  
    This tells GitHub how to build your cloud dev environment.                                    
                       
    Example:
                                                                                                  
      ─ json           
      {
        "name": "Learning Lab",
        "image": "mcr.microsoft.com/devcontainers/base:ubuntu",
        "features": {
          "ghcr.io/devcontainers/features/node:1": {
            "version": "lts"                                                                      
          },                                                                                      
          "ghcr.io/devcontainers/features/python:1": {
            "version": "3.12"                                                                     
          }                                                                                       
        },                                                                                        
        "customizations": {                                                                       
          "vscode": {  
            "extensions": [
              "ms-python.python",
              "ms-toolsai.jupyter",                                                               
              "dbaeumer.vscode-eslint",
              "esbenp.prettier-vscode"                                                            
            ]                                                                                     
          }                                                                                       
        }                                                                                         
      }                
                                                                                                  
    That means every time you create a Codespace, it can automatically have:                      
                                                                                                  
    • Python                                                                                      
    • Node.js                                                                                     
    • VS Code extensions                                                                          
    • project dependencies
    • terminal tools                                                                              
                                                                                                  
    That’s very DevOps-like: environment as code.

## Monthly Calendar

### Static Calendar

```python
import calendar
from IPython.display import HTML, display

year = 2026
month = 7

cal = calendar.HTMLCalendar()
display(HTML(cal.formatmonth(year, month)))
```
