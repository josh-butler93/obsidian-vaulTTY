#### **1. Inputs**

```python
#inputs
num1 = int(input("Numb 1: "))
num2 = int(input("Numb 2: "))
#process
if num1 > num2:
    print(num1, ">", num2)
elif num2 > num1:
    print(num2, ">", num1)
else:
    print(num1, "=", num2)
#output
```

#### **2. Start Stop**

```python
#input
total = 0

start = int(input("Start: "))
stop = int(input("Stop: ")) +1
for x in range(start, stop):
    total += x

print("Total: ", total)
#process

#output
```

#### **3. Seperated Values**

```python
#Ex3_3
nums = input("Comma seperated numbers: ")

listNums = nums.split(",")
for num in listNums:
    if int(num) > 0:
      print(num)
```
